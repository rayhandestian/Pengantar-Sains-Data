```python
import pandas as pd
import numpy as np

file_path = "datasets/volume-pengangkutan-sampah.csv"

# Load dataset, ';' is delimiter
df = pd.read_csv(file_path, delimiter=';')

# Sample
df.head()
```

```python
# Dataset info
df.info()
```

```python
# Display unique values in satuan_panjang/luas
print("Unique units before cleaning:", df['satuan_panjang/luas'].unique())

# Fill missing satuan_panjang/luas' w/ 'm' asumsi meter
df['satuan_panjang/luas'].fillna('m', inplace=True)

# Verify
print("Unique units after cleaning:", df['satuan_panjang/luas'].unique())
```

```python
# Replace '-' with NaN
df['volume_sampah_perhari(m3)'] = df['volume_sampah_perhari(m3)'].replace('-', np.nan)

# Convert de column to float
df['volume_sampah_perhari(m3)'] = pd.to_numeric(df['volume_sampah_perhari(m3)'], errors='coerce')

# Check missing values
df['volume_sampah_perhari(m3)'].isnull().sum()
```

```python
# Calculate mean volume per kecamatan
mean_volume = df.groupby('kecamatan')['volume_sampah_perhari(m3)'].transform('mean')

# Isi missing values dengan volume rata-rata (mean) from masing2 kecamatan bcz it's easy
df['volume_sampah_perhari(m3)'].fillna(mean_volume, inplace=True)

# Verify no missing values
df['volume_sampah_perhari(m3)'].isnull().sum()
```

```python
# Calculate mean panjang/luas for each kecamatan
mean_panjang_per_kecamatan = df.groupby('kecamatan')['panjang/luas'].transform('mean')

# Isi missing values dengan volume rata-rata (mean) from masing2 kecamatan yeah
df['panjang/luas'].fillna(mean_panjang_per_kecamatan, inplace=True)

# Check how many missing values remain
missing_panjang = df['panjang/luas'].isnull().sum()
print(f"Missing 'panjang/luas' after filling per 'kecamatan': {missing_panjang}")
```

```python
# WHAT
# Calculate the overall mean of 'panjang/luas'
overall_mean_panjang = df['panjang/luas'].mean()

# Isi missing values sisanya dengan rata-rata overall
df['panjang/luas'].fillna(overall_mean_panjang, inplace=True)

# No more missing values?
missing_panjang_final = df['panjang/luas'].isnull().sum()
print(f"Missing 'panjang/luas' after filling overall mean: {missing_panjang_final}")
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Convert bulan dan tanggal ke integer
df['bulan'] = df['bulan'].astype(int)
df['tanggal'] = df['tanggal'].astype(int)

# Convert panjang/luas ke float
df['panjang/luas'] = pd.to_numeric(df['panjang/luas'], errors='coerce')

# Verify
df.dtypes
```

```python
# Display the cleaned DataFrame
df
```

```python
# Define output file path
output_file_path = "datasets/volume-pengangkutan-sampah-cleaned.csv"

# Save cleaned DataFrame to new CSV file
df.to_csv(output_file_path, index=False, sep=';')

print(f"Cleaned dataset saved to {output_file_path}")
```
